# Text Cleaning for Machine Learning

Text cleaning is an essential preprocessing step for any natural language processing (NLP) or machine learning task involving text. This document outlines the key steps, considerations, and Python code examples for effective text cleaning.

---

## Purpose of Text Cleaning

- Reduce noise in text data  
- Normalize input for consistency  
- Improve model performance and generalization  

---

## Common Text Cleaning Operations

| Operation                 | Description |
|--------------------------|-------------|
| **Lowercasing**          | Convert all text to lowercase to unify vocabulary (`"The" → "the"`) |
| **Punctuation Removal**  | Remove characters like `.,!?` unless needed for sentiment or syntax |
| **Digit Removal or Replacement** | Replace numbers with `<NUM>` or remove them if irrelevant |
| **Whitespace Trimming**  | Remove extra spaces, tabs, and newline characters |
| **Stopword Removal**     | Remove common words like `is`, `the`, `and` |
| **Stemming / Lemmatization** | Normalize words to their base forms (`"running" → "run"`) |
| **HTML Tag Removal**     | Clean raw text scraped from websites |
| **Emoji & Symbol Removal** | Remove emojis if not needed for analysis |
| **Regex Cleaning**       | Use patterns to remove emails, URLs, dates, etc. |


## When to Clean and What to Keep

| Task            | Special Consideration |
|-----------------|------------------------|
| Sentiment Analysis | Keep emojis, punctuation like `!`, `?` |
| Text Classification | Normalize as much as possible |
| Translation      | Preserve structure and meaning |
| Chatbots         | Handle contractions (`I'm → I am`), slang, and spelling errors |

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
```

```python
def clean_text(text):
    print("Original:", text)

    text = text.lower()
    print("Lowercase:", text)

    text = re.sub(r"http\S+", "", text) # sub means substitute
    print("Remove URLs:", text)

    text = re.sub(r"<.*?>", "", text)
    print("Remove HTML:", text)

    text = re.sub(r"[^a-z\s]", "", text)
    print("Remove punctuation & digits:", text)

    tokens = text.split()
    print("Tokenized:", tokens)

    stop_words = set(stopwords.words("english"))
    tokens = [t for t in tokens if t not in stop_words]
    print("Remove stopwords:", tokens)

    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(t) for t in tokens]
    print("Lemmatized:", tokens)

    cleaned = " ".join(tokens)
    print("Final:", cleaned)

    return cleaned


text = "This is a <b>TEST</b> sentence with a URL: http://example.com, numbers 123, and emojis 😊."
clean_text(text)
```

## Common Pitfalls

- **Over-cleaning**: Removing too much can strip meaning  
- **Under-cleaning**: Noisy data leads to poor model performance  
- **Language-specific stopwords**: Ensure stopword lists match your language  
- **Domain-specific vocab**: Words like "COVID-19" may be important  

---

## Final Tips

- Always validate cleaning steps with a small data sample  
- Customize cleaning per task and domain  
- Log cleaning steps for reproducibility
