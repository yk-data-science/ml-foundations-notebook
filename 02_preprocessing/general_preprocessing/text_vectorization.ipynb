## Text Vectorization in Machine Learning

Text vectorization is the process of converting textual data into numerical representations so that machine learning algorithms can understand and work with them. It is a critical step in NLP pipelines.

### Common Techniques

- **Bag of Words (BoW)**  
  Represents text by word count. Ignores word order but preserves frequency.

- **TF-IDF (Term Frequency - Inverse Document Frequency)**  
  Balances word frequency with how unique the word is across documents.

- **Word Embeddings (e.g., Word2Vec, GloVe)**  
  Captures semantic meaning of words in dense vectors.

- **Contextual Embeddings (e.g., BERT)**  
  Uses deep learning to understand context.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Sample corpus
corpus = [
    "Machine learning is interesting",
    "Natural language processing with machine learning",
    "Text vectorization converts words to numbers"
]

# Bag of Words - Count Vectorization
bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(corpus)
print("Bag of Words Matrix:\n", bow_matrix.toarray())
print("Vocabulary:\n", bow_vectorizer.vocabulary_)

# TF-IDF - Term Frequency-Inverse Document Frequency
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)
print("\nTF-IDF Matrix:\n", tfidf_matrix.toarray())
print("Vocabulary:\n", tfidf_vectorizer.vocabulary_)
```

## Word Embeddings and Contextual Embeddings

### Word Embeddings with Word2Vec

```python
%pip install gensim
```

```python
from gensim.models import Word2Vec

# Sample corpus
sentences = [
    ["machine", "learning", "is", "fun"],
    ["natural", "language", "processing"],
    ["word", "embedding", "captures", "meaning"]
]

# Train Word2Vec model
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, workers=4)

# Get vector for a word
vector = model.wv["machine"]
print("Vector for 'machine':\n", vector)

# Find similar words
similar = model.wv.most_similar("learning")
print("Words similar to 'learning':\n", similar)
```

### Contextual Embeddings with BERT

```python
%pip install transformers
```

```python
from transformers import BertTokenizer, BertModel
import torch

# Load BERT base model and tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')
```

```python
# Input sentence
sentence = "Machine learning is powerful"

# Tokenize input
inputs = tokenizer(sentence, return_tensors="pt") # pt for PyTorch

# Get BERT embeddings
with torch.no_grad():
    outputs = model(**inputs)

# Extract the last hidden state (token-level embeddings)
last_hidden_states = outputs.last_hidden_state
print("Shape of token embeddings:", last_hidden_states.shape)  # (batch_size, seq_len, hidden_dim)

# Get sentence embedding (e.g., mean pooling)
sentence_embedding = last_hidden_states.mean(dim=1)
print("Sentence embedding:\n", sentence_embedding)
```

```python
import torch.nn.functional as F

model.eval()  # Set the model to evaluation mode

# Two example sentences
sent1 = "Machine learning is powerful"
sent2 = "Artificial intelligence is strong"

def get_sentence_embedding(sentence):
    # Tokenize input sentence
    inputs = tokenizer(sentence, return_tensors="pt")
    # Get model output without gradient calculation
    with torch.no_grad():
        outputs = model(**inputs)
    # Extract last hidden states (token embeddings)
    last_hidden = outputs.last_hidden_state
    # Return mean pooled sentence embedding
    return last_hidden.mean(dim=1)

# Get sentence embeddings
emb1 = get_sentence_embedding(sent1)
emb2 = get_sentence_embedding(sent2)

# Calculate cosine similarity between embeddings
similarity = F.cosine_similarity(emb1, emb2)

print(f"Cosine Similarity: {similarity.item():.4f}")
```

## When to Use What?

| Method         | Use Case                                | Pros                          | Cons                        |
|----------------|------------------------------------------|-------------------------------|-----------------------------|
| BoW            | Simple models, small datasets            | Easy to implement             | Sparse, ignores context     |
| TF-IDF         | Baseline text classification             | Weighs rare terms             | Still sparse, no semantics  |
| Word2Vec/GloVe | Semantic similarity, word-level tasks     | Dense, captures meaning       | Needs pretraining/loading   |
| BERT/ELMo      | Context-sensitive tasks (e.g., QA, NER)   | State-of-the-art performance  | Heavy computation required  |


## Notes

- Always clean and normalise text before vectorizing.
- Consider dimensionality reduction (e.g., PCA, TruncatedSVD) if feature size is too large.
- For deep learning models, use embeddings instead of BoW/TF-IDF.
